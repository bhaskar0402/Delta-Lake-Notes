# Delta Table Optimization Notebook

This notebook focuses on key Delta Lake maintenance operations:

- **Optimizing Tables**: Use the OPTIMIZE command to compact small files and improve query performance.
- **ZORDER Optimization**: Apply ZORDER BY to colocate related data, enabling efficient data skipping and faster queries.
- **Running VACUUM**: Remove obsolete files and reclaim storage space by executing the VACUUM command.

These features help maintain efficient, performant Delta tables in your data lake.

```python
%sql
drop table if exists training.delta_demo.people_delta_demo
```

```python
from pyspark.sql import Row

# Create initial DataFrame with sample data
data = [Row(id=1, name='Alice', age=30), Row(id=2, name='Bob', age=25), Row(id=3, name='Charlie', age=35)]
df = spark.createDataFrame(data)

# Overwrite and create a managed Delta table with the DataFrame in the training.delta_demo schema
df.write.format('delta').mode('overwrite').saveAsTable('dev.delta.people_delta_demo')

# Read the Delta table to verify creation and display its contents
delta_df = spark.read.table('dev.delta.people_delta_demo')
display(delta_df)
```

```python
df.write.format('delta').mode('append').save("/Volumes/dev/default/delta/mytable")
```

```python
df2=spark.read.format('delta').load("/Volumes/dev/default/delta/mytable")
display(df2)
```

```python
%sql
optimize "/Volumes/dev/default/delta/mytable"
```

```python
%sql
optimize dev.delta.people_delta_demo
```

```python
%sql
optimize "/Volumes/dev/default/delta/mytable" zorder by id
```

```python
%sql
optimize dev.delta.people_delta_demo zorder by id
```

```python
%sql
DESCRIBE DETAIL dev.delta.people_delta_demo;
```

```python
%sql
vacuum dev.delta.people_delta_demo dry run 
```
